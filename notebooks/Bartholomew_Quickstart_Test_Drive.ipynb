## Step 3: Protecting Agent Tools with @guard.protect

```python
@guard.protect
def execute_database_query(query: str):
    return f"[SUCCESS] Query executed: {query}"

# 1. Benign query
print("--- 1. Permitted Safe Action ---")
result = execute_database_query(query="SELECT id, username, email FROM users WHERE tenant_id = 101;")
print(result)

# 2. Malicious destructive query
print("\n--- 2. Blocked Destructive Action ---")
try:
    execute_database_query(query="DROP TABLE users CASCADE;")
except Exception as e:
    print(f"[BLOCKED] Gated by Bartholomew Invariant Engine:\n{e}")
```

## Step 4: Zero-Leak Credential and Secret Scrubbing (<100us)

```python
raw_text = "Connecting with Bearer sk-ant-api03-abcdef1234567890abcdef1234567890-test and AWS key AKIAIOSFODNN7EXAMPLE"
sanitized_text = guard.scrub(raw_text)

print("Raw Input:      ", raw_text)
print("Scrubbed Output:", sanitized_text)
```

## Step 5: Cryptographic Audit Receipts (RFC 8785 Canonical JSON)

```python
# Execute pre-flight check and retrieve signed receipt
check_result = guard.check("git status --short")
print("Verdict:        ", check_result.get("verdict"))
print("Latency (us):   ", check_result.get("latency_us"))
print("Receipt SHA-256:", check_result.get("receipt_sha256"))
print("Signed Receipt: ", check_result.get("receipt"))
```

## Step 6: Palantir AIP and Model Context Protocol (MCP) Protection

```python
# Evaluate an arbitrary code payload intended for an MCP server
mcp_payload = "import os; os.system('cat /etc/shadow')"
eval_result = guard.evaluate_ast(mcp_payload, language="python")

print("MCP Tool Call Evaluated:")
print(f"  Allowed:    {eval_result.get('allowed')}")
print(f"  Violations: {eval_result.get('violations')}")
print(f"  Latency:    {eval_result.get('latency_us')} us")
```
